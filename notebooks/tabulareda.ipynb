```python
import os
import glob

parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
txt_files = glob.glob(os.path.join(parent_dir, "data/googledeepmind/math/train-easy/*.txt"))

total_pairs = 0
file_counts = []

# Output an onverview on a subset of the data
print("Math Dataset train-easy Overview:")
for file_path in txt_files:
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        num_pairs = len(lines) // 2
        total_pairs += num_pairs
        file_counts.append((os.path.basename(file_path), num_pairs))

# Display each file and count iteratively
for fname, count in sorted(file_counts):
    print(f"{fname}: {count:,} Q&A pairs")

print(f"\nTotal Q&A pairs across {len(file_counts)} files: {total_pairs:,}")
```
