```python
import os
import glob


parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir)) 
txt_files = glob.glob(os.path.join(parent_dir, "data/googledeepmind/math/train-easy/*.txt"))[:10]

for file_path in txt_files:
    with open(file_path, 'r', encoding='utf-8') as file:
        contents = file.read()
        print(f"Contents of {file_path}:")
        print(contents)
        print("\n" + "-"*40 + "\n")
```

```python
parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir)) 
txt_files = glob.glob(os.path.join(parent_dir, "data/googledeepmind/math/train-easy/*.txt"))[:100]

data_samples = []
for file_path in txt_files:
    with open(file_path, 'r', encoding='utf-8') as file:
        contents = file.read()
        data_samples.append({
            "filename": os.path.basename(file_path),
            "content": contents
        })
```

```python
import pandas as pd

df = pd.DataFrame(data_samples)
df.head()
```
