```python
import os
import glob

parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
txt_files = glob.glob(os.path.join(parent_dir, "data/googledeepmind/math/train-easy/*.txt"))

total_pairs = 0
file_counts = []

# Output an onverview on a subset of the data
print("Math Dataset train-easy Overview:")
for file_path in txt_files:
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
        num_pairs = len(lines) // 2
        total_pairs += num_pairs
        file_counts.append((os.path.basename(file_path), num_pairs))

# Display each file and count iteratively
for fname, count in sorted(file_counts):
    print(f"{fname}: {count:,} Q&A pairs")

print(f"\nTotal Q&A pairs across {len(file_counts)} files: {total_pairs:,}")
```

```python
import os

file_path = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
def verify_qna_pairs(file_path, sample_n=5):
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
    
    if len(lines) % 2 != 0:
        print(f"Warning: Odd number of lines in {file_path} — possible formatting issue.")
    else:
        print(f"Total lines: {len(lines)} — valid number of Q&A pairs: {len(lines)//2}")
    
    print(f"\nFirst {sample_n} Q&A pairs from {os.path.basename(file_path)}:\n")
    for i in range(0, min(2 * sample_n, len(lines)), 2):
        question = lines[i]
        answer = lines[i + 1]
        print(f"Q{i//2 + 1}: {question}\nA{i//2 + 1}: {answer}\n")
        
test_file = os.path.join(file_path, "data/googledeepmind/math/train-easy/algebra__linear_1d.txt")
verify_qna_pairs(test_file)
```

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt

# Set file paths
parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
data_dir = os.path.join(parent_dir, "data/googledeepmind/math/train-easy")
txt_files = glob.glob(os.path.join(data_dir, "*.txt"))[:5]  # Limit for speed

# Load a few math test samples
samples = []

for file_path in txt_files:
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
        for i in range(0, min(100, len(lines) - 1), 2):  # max 50 pairs per file
            question = lines[i]
            answer = lines[i + 1]
            samples.append({
                "module": os.path.basename(file_path),
                "question": question,
                "answer": answer,
                "q_len": len(question),
                "a_len": len(answer)
            })

df = pd.DataFrame(samples)
print(f"Loaded {len(df)} Q&A samples from {len(txt_files)} files.")

# Generate a few plots
plt.figure(figsize=(12, 5))

# Question length distribution
plt.subplot(1, 2, 1)
df["q_len"].hist(bins=30)
plt.title("Question Length Distribution")
plt.xlabel("Length (characters)")
plt.ylabel("Frequency")

# Answer length distribution
plt.subplot(1, 2, 2)
df["a_len"].hist(bins=30, color='orange')
plt.title("Answer Length Distribution")
plt.xlabel("Length (characters)")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
