```python
from datasets import get_dataset_config_names

# Load dataset configs
omega_explorative = "allenai/omega-explorative"
omega_transformative = "allenai/omega-transformative"
omega_compositional = "allenai/omega-compositional"

# Pull configs individually
configs_explorative = get_dataset_config_names(omega_explorative)
configs_transformative = get_dataset_config_names(omega_transformative)
configs_compositional = get_dataset_config_names(omega_compositional)

# Append configs into a single list or dict
all_configs = {
    "explorative": configs_explorative,
    "transformative": configs_transformative,
    "compositional": configs_compositional
}

# Generate Results
for k, v in all_configs.items():
    print(f"{k}: {v}\n")
```

```python
import logging
import datasets
import pandas as pd
from datasets import load_dataset
from concurrent.futures import ThreadPoolExecutor

# Logging configs
#datasets.logging.set_verbosity(0)
#logging.getLogger("datasets").setLevel(logging.ERROR)
#logging.getLogger("datasets.builder").setLevel(logging.ERROR)
#logging.getLogger("datasets.download").setLevel(logging.ERROR)

# Set to True for more logging info
#VERBOSE = False

# Set dataset configs
all_datasets = {
    "explorative": "allenai/omega-explorative",
    "transformative": "allenai/omega-transformative",
    "compositional": "allenai/omega-compositional"
}

all_configs = {
    "explorative": configs_explorative,
    "transformative": configs_transformative,
    "compositional": configs_compositional
}

load_targets = [
    (all_datasets[dataset_key], config)
    for dataset_key, config_list in all_configs.items()
    for config in config_list
]

# Set configs and load data
def load_and_tag_config(target):
    dataset_name, config = target
    print(f"Loading {dataset_name} / config: {config}")
    ds = load_dataset(dataset_name, config, split="train")
    df = pd.DataFrame(ds)
    df["module"] = config
    df["dataset"] = dataset_name
    return df

    
# Load in parallel
with ThreadPoolExecutor() as executor:
    dfs = list(executor.map(load_and_tag_config, load_targets))

# Concatenate. Generate results
dataframe = pd.concat(dfs, ignore_index=True)
print(f"\nTotal rows loaded: {len(dataframe)}")
print(f"Modules loaded: {len(dfs)}")
```
